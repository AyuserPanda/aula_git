## Tópico 1: Funções em Python (`def`)

As funções nos permitem modularizar o código, reutilizar trechos de lógica e manter a organização do pipeline de tratamento de dados.

### 💡 Exemplo da Sala de Aula:

````python
def calcular_imposto(valor: float, taxa: float) -> float:
    """
    Calcula o imposto sobre um valor com uma taxa padrão de 15%.
    Retorna o valor final com o imposto aplicado.
    """
    imposto = valor * taxa
    valor_final = valor + imposto
    return valor_final

# Usando a função:
preco_notebook = calcular_imposto(3500.00, 0.15)
print(f"Preço final: R$ {preco_notebook}")
````

## Tópico 2: Funções Lambda e Aplicações com `map()`

Funções `lambda` são funções anônimas de linha única perfeitas para transformações rápidas de dados.

### 💡 Exemplo da Sala de Aula:

````python
# Função Clássica
def quadrado(x):
    return x ** 2

# Mesma função, versão Lambda
quadrado_lambda = lambda x: x ** 2

print(quadrado_lambda(5)) # Saída: 25

# APLICAÇÃO REAL: Usando Lambda com a função map() para aplicar a regra em uma lista inteira
precos_sujos = [10.50, 20.00, 35.99]
precos_com_desconto = list(map(lambda p: p * 0.90, precos_sujos))

print(precos_com_desconto) # Saída: [9.45, 18.0, 32.391]
````

### ✏️ Exercício 2.1: Conversão de Temperaturas com Lambda e Map
Dada a lista de temperaturas em Celsius: `temperaturas_celsius = [0.0, 18.5, 25.0, 32.2, 40.0]`
1. Use `map()` e uma expressão `lambda` para converter todas as temperaturas para Fahrenheit.
   *(Fórmula: F = C * 1.8 + 32)*
2. Armazene o resultado em uma lista e imprima o resultado.

````python
# SEU CÓDIGO AQUI:
temperaturas_celsius = [0.0, 18.5, 25.0, 32.2, 40.0]

temperaturas_fahrenheit = list(map(lambda c: c * 1.8 + 32, temperaturas_celsius))

for c, f in zip(temperaturas_celsius, temperaturas_fahrenheit):
    print(f"{c}°C = {f:.1f}°F")
````

## 🔍 Tópico 3: Expressões Regulares (REGEX)

Expressões regulares são fundamentais no processo de Engenharia de Dados para buscar, extrair e higienizar textos não estruturados.

### 💡 Exemplos da Sala de Aula:

#### Exemplo 1: Extração e Limpeza de Preços em Atendimento

````python
import re

texto_chat = """
Atendente: O notebook gamer custa R$ 4.500,00 na loja física.
Cliente: Mas no site vi por R$3999,90!
Atendente: Consigo fazer por R$ 4.000,00 se pagar à vista.
"""

# 1. Encontrando os padrões de dinheiro no texto
padrao_dinheiro = r'R\$\s*[\d\.,]+'
precos_brutos = re.findall(padrao_dinheiro, texto_chat)

print("Preços encontrados no chat:", precos_brutos)
# Saída: ['R$ 4.500,00', 'R$3999,90', 'R$ 4.000,00']

# 2. Transformando em Float (Opcional, mas nível Sênior!)
precos_matematicos = []
for p in precos_brutos:
    # Remove R$, espaços e pontos de milhar, e troca a vírgula por ponto decimal
    p_limpo = re.sub(r'[R\$\s\.]', '', p).replace(',', '.')
    precos_matematicos.append(float(p_limpo))

print("Preços prontos para o Numpy:", precos_matematicos)
````

#### Exemplo 2: Limpeza de CPF e Telefone

````python
import re

texto_cliente = "Nome: João, Contato: (11) 98888-7777, CPF: 123.456.789-00!"

# 1. Extraindo apenas os números do CPF (Limpando pontuação)
cpf_limpo = re.sub(r'\D', '', "123. 456.789-00") # Substitui o que não for número por VAZIO ('')
print("CPF Limpo:", cpf_limpo) # Saída: 12345678900

# 2. Extraindo o telefone ignorando o texto ao redor
padrao_telefone = r'\(\d{2}\)\s\d{5}-\d{4}' # Busca: (XX) XXXXX-XXXX
telefone = re.findall(padrao_telefone, texto_cliente)
print("Telefone Extraído:", telefone)
````

#### Exercício - Relatório de Estoque (Uso de Grupos de Captura)

````python
import re

relatorio_estoque = """
Registro A: Lote-8849 - TV 50 Polegadas - Saldo: 45 unidades - Valor: 3.50
Registro B: Lote-1023 - Geladeira Frost Free - Saldo: 12 unidades - Valor: 20.00
Registro C: Lote-9921 - Micro-ondas - Saldo: 8 unidades - Valor: 30.60
"""
````

````python
# DESAFIO 1: Extrair apenas os números dos Lotes

#DEFINA padrao_lote

padrao_lote = r"Lote-(\d+)"

lotes_extraidos = re.findall(padrao_lote, relatorio_estoque)
print("Números dos Lotes (IDs):", lotes_extraidos)
````

````python
# DESAFIO 2: Extrair apenas as quantidades do saldo

padrao_saldo = r"Saldo: (\d+) unidades"

saldos_extraidos = re.findall(padrao_saldo, relatorio_estoque)
print("Quantidades em Estoque:", saldos_extraidos)
````

````python
# DESAFIO 3: Extrair apenas os valores unitários

padrao_valor = r"Valor: (\d+\.\d+)"

valor_extraidos = re.findall(padrao_valor, relatorio_estoque)
print("Valor Unitario:", valor_extraidos)
````

### ✏️ Exercício 3.4: Extração de Domínio de E-mails com Regex (re).
Instruções: Dado o texto contendo logs de acesso:  
`logs = "Usuários ativos: ana.silva@empresa.com.br, carlos@gmail.com, suporte@tecnologia.org"`    
1. Escreva um padrão Regex para extrair apenas o nome do domínio de cada e-mail (ou seja, apenas a parte que fica depois do @).  
2. Imprima a lista com os domínios encontrados

Vamos analisar os e-mails para entender o padrão:

```
ana.silva@empresa.com.br
carlos@gmail.com
suporte@tecnologia.org
```

Queremos capturar só a parte **depois do @** (`empresa.com.br`, `gmail.com`, `tecnologia.org`).

## Um símbolo novo: `\w`

Além do `\d` (dígito) que já usamos, agora precisamos de um símbolo que capture **letras, números e pontos** (porque domínios têm letras e pontos, tipo `empresa.com.br`):

| Símbolo | Significado |
|---|---|
| `\w` | Qualquer "caractere de palavra" (letras, números e underscore `_`) |
| `\w+` | Um ou mais desses caracteres seguidos |

Mas repare que domínio tem **pontos** também (`.com.br`), e `\w` sozinho não inclui ponto. Por isso, vamos criar um "grupo de caracteres permitidos" usando colchetes `[ ]`:

```
[\w.-]+
```
Isso significa: "um ou mais caracteres que podem ser letra, número, ponto (`.`) ou hífen (`-`)".

## Código

```python
import re

logs = "Usuários ativos: ana.silva@empresa.com.br, carlos@gmail.com, suporte@tecnologia.org"

padrao_dominio = r"@([\w.-]+)"

dominios_extraidos = re.findall(padrao_dominio, logs)
print("Domínios encontrados:", dominios_extraidos)
```

### Resultado esperado:
```
Domínios encontrados: ['empresa.com.br', 'gmail.com', 'tecnologia.org']
```

## Explicando o padrão `r"@([\w.-]+)"`

- **`@`** → texto literal, procura exatamente o símbolo arroba
- **`(...)`** → captura o que estiver dentro dos parênteses
- **`[\w.-]`** → um "conjunto" de caracteres aceitos: letras/números (`\w`), ponto (`.`) ou hífen (`-`)
- **`+`** → repete esse conjunto uma ou mais vezes seguidas

### Por que colocar `-` dentro do colchete?
Alguns domínios têm hífen (tipo `minha-empresa.com`), então incluímos por segurança, mesmo que não apareça nesse exemplo específico.

### Um detalhe técnico sobre a posição do hífen
Dentro dos colchetes `[ ]`, o hífen (`-`) tem um significado especial quando fica **no meio** (tipo `[a-z]` significa "de a até z"). Por isso, colocamos ele **no final** do grupo (`[\w.-]`), onde ele é interpretado como um hífen literal, sem confundir com uma "faixa" de caracteres.
